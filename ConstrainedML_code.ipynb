### R code for the likelihood function in Constrained ML
Below is the log-likelihood (Eqn 7) in terms of the transformed coordinates ${u, v, w}$. The first argument <code>parms</code> is a vector of length three, storing ${u, v, w}$ (order matters), and the second argument <code>n</code> is a vector of the input nine genotypes. 

I then maximise the log-likelihood w.r.t. ${u, v, w}$ via the built-in <code>optim()</code>. I'm sure there are other rountines avilable, on R and elsewhere. L-BFGS-B allows me to set the cube boundaries. Finally we transform the solutions back to frequency domain. 

```r
# THE LOG-LIKELIHOOD (EQN 7) IN THE NEW COORDINATES
loglike<-function(parms, n)
{
    # THE NEW CUBE COORDINATES
    u<-parms[1]
    v<-parms[2]
    w<-parms[3]
    # FREQ COORDINATES
    pAB<-u*v*w
    pAb<-u*v*(1-w)
    paB<-u*(1-v)
    pab<-1-u
    # EXPECTED FREQ
    f<-c(pAB^2, 2*pAB*pAb, pAb^2, 
         2*pAB*pAb, 2*(pAB*pab+pAb*paB), 2*pAb*pab, 
         paB^2, 2*paB*pab, pab^2)
    # MULTINOMIAL LOG-LIKE
    return(sum(n*log(f)))
}
```

```r
# THE DATA (NINE GENOTYPE COUNTS)
n<-c(10, 3, 4, 2, 8, 6, 7, 8, 9)
```

```r
# MAXIMISE VIA THE BUILT-IN optim()
m<-optim(rep(1/2, 3), fn=loglike, n=n, 
      method='L-BFGS-B', 
      lower=rep(1e-10, 3), upper=rep(1-1e-10, 3),
      control=list(fnscale=-1))
m
```

```r
# BACK TRANSFORM TO FREQ
f<-c(m$par[1]*m$par[2]*m$par[3], m$par[1]*m$par[2]*(1-m$par[3]), m$par[1]*(1-m$par[2]), 1-m$par[1])
f
sum(f)
```
