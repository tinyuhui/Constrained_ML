## Notes on Constrained ML

#### 1. Constrained ML (estimating haplotype frequencies from unphased genotypes)
One contribution of this paper is provide a new algorithm to estimate haplotype frequencies from unpahsed genotypes for a pair of biallelic loci. It is not a new problem, however, it has always been a problem. 

Given the counts of the nine genotypes, the two configurations of the double heterozygotes cannot be differentiated - we only know their sum (middle cell, Table 1). For instance, the observed count for haplotype AB $=2*AABB + AABb + AaBB + (some~of~AaBb)$. To tackle the problem, we first establish the expected genotypic frequencies under HWE (Table 1), then construct the log-likelihood function under multinomial sampling (Eqn 7): 

$$ l(p_{AB}, p_{Ab}, p_{aB})=constant+\sum_{i=1}^9{n_i*log(f_i)}$$

where $n_i$ and $f_i$ are the observed counts and expected frequencies of the nine genotypes. The parameter space is trivariate, as the four haplotype frequency is redundant. Note that there are other constraints as they are all frequencies. 

Excoffier and Slatkin (1995) and Gaunt et al. (2007) both maximise the same log-likelihood w.r.t. haplotype frequencies (i.e. all three methods should give identical results). The constrints are that they are frequencies and are bounded between 0 and 1, and they should sum to 1. To my knowledge these constaints make optimisation difficult, and/or inefficient, and some solutions may be biologically infeasible. That's why I proposed transforming the parameter space into a set of coordinates, then maximising the log-likelihood within a cube, before back-transforming the solution into frequencies. The transformation is as follows: 

$$u=p_{AB}+p_{Ab}+p_{aB}$$
$$v=\frac{p_{AB}+p_{Ab}}{p_{AB}+p_{Ab}+p_{aB}}$$
$$w=\frac{p_{AB}}{p_{AB}+p_{Ab}}$$

And now ${u, v, w}$ forms a unit cube. The backward transform is: 

$$p_{AB}=u*v*w$$
$$p_{Ab}=u*v*(1-w)$$
$$p_{aB}=u*(1-v)$$

This method should give better and faster convergence than EM. Besides, optimising a multidimensional function within a box-like constraint is a well-studied problem with many routines avaible in R and Python (say), so users do not have to reinvent it or write their own EM. I think it is a neat and beautiful solution. CubeX is beautiful as well, note that CubeX requires to you rotate/transpose your nine genotypes in a specific order (largest counts on the top-left?) to calculate the coefficients. 

The same transformation applies to multiallelic loci as hyper-cubes (see Additional File).

#### 3. Sample size correction for $r^2$
Another (minor) contribution of the paper is the sample size correction for $r^2$. It follows that if the true $r^2=1$ then the sampled $r^2$ should also be 1 (this measure cannot exceed 1, given the nine genotypes). This is inspired by Park (2012), which to me is a hidden gem.
